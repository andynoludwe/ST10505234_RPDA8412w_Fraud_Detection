# Detecting AI use on bank documents: synthetic-data project
Run from the `ai_doc_forensics` folder. All documents, signatures and metadata are synthetic (fictional bank and employer). Read `README.md` for the assumptions behind the results.

```
import sys, subprocess
subprocess.run([sys.executable, '-m', 'pip', 'install', '-r', 'requirements.txt'])
```

```
# Quick trial (about 2 minutes). For the full study use: --n_cases 1800 --difficulty all
r = subprocess.run([sys.executable, 'run_pipeline.py', '--n_cases', '600', '--difficulty', 'medium', '--out', 'runs_trial'], capture_output=True, text=True)
print(r.stdout[-3000:]); print(r.stderr[-1500:])
```
